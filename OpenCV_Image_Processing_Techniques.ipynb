```python
#image slicing
img=img[20:70,100:200]
img
```

```python
#image rotating
img2=cv2.rotate(img2,cv2.ROTATE_90_CLOCKWISE)
```

```python
from google.colab.patches import cv2_imshow
cv2_imshow(img2)
```

```python
img2=cv2.rotate(img2,cv2.ROTATE_180)
```

```python
from google.colab.patches import cv2_imshow
cv2_imshow(img2)
```

```python
#image fliping
img2=cv2.flip(img2,1)
```

```python
from google.colab.patches import cv2_imshow
cv2_imshow(img2)
```

```python
#resizing
img2=cv2.resize(img,(100,100))
```

```python
from google.colab.patches import cv2_imshow
cv2_imshow(img2)
```

```python
#image bluring
img2=cv2.blur(img,(23,50))
```

```python
from google.colab.patches import cv2_imshow
cv2_imshow(img2)
```

```python
gray_image=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
edges=cv2.Canny(gray_image,100,200) # Canny is edge reduction
from google.colab.patches import cv2_imshow
cv2_imshow(img)
cv2_imshow(edges)
```
